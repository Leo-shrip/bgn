```python
import re

def parse_events_string(event_string):
    """
    解析事件字符串，返回一个包含(时间戳, 事件类型)元组的列表。
    示例输入: "8:24742NT_START,24743RSOUND,24743RLICK..."
    """
    events = []
    # 移除开头的编号（如"12:"）
    event_string = re.sub(r'^\d+:', '', event_string)
    # 匹配规则：时间戳（数字） + 事件类型（非逗号字符）
    pattern = r'(\d+)(\w+)'
    matches = re.findall(pattern, event_string)
    
    for timestamp, event_type in matches:
        events.append((int(timestamp), event_type))
    
    return events

# 应用解析函数，为每个试次创建一个新列'parsed_events'
expanded_data['parsed_events'] = expanded_data['events'].apply(parse_events_string)
```

```python
# 3. 创建新的DataFrame，将嵌套列表“炸开”成多行
# 每行代表一个独立的事件记录
events_rows = []
for idx, row in expanded_data.iterrows():
    trial_num = row['trial']
    for timestamp, event_type in row['parsed_events']:
        events_rows.append({
            'trial': trial_num,
            'event_type': event_type,
            'timestamp': timestamp
        })
# 转换为DataFrame
events_expanded_df = pd.DataFrame(events_rows)
events_expanded_df
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 假设您的DataFrame名为 df
# 1. 定义一个函数来处理单个trial的数据，将其转换为相对时间
def align_trial_to_sound(trial_row):
    """
    将单个trial的事件时间对齐到该trial的声音刺激发生时间。
    逻辑：优先使用LSOUND，如果LSOUND为空则使用RSOUND作为时间零点。
    """
    # 确定基准时间 (t0)
    # 优先使用LSOUND列表的第一个元素，如果LSOUND列表为空，则使用RSOUND列表的第一个元素
    if trial_row['LSOUND'] and len(trial_row['LSOUND']) > 0:
        t0 = trial_row['LSOUND'][0]
    elif trial_row['RSOUND'] and len(trial_row['RSOUND']) > 0:
        t0 = trial_row['RSOUND'][0]
    else:
        # 如果该trial没有声音事件，则跳过（返回空列表）
        return None

    # 定义一个时间窗口（例如，声音前1秒到声音后3秒）
    time_window_start = -1000  # 单位需与你的时间戳一致（假设是毫秒）
    time_window_end = 3000

    # 初始化一个列表，用于存储这个trial内，在时间窗里发生的所有事件信息
    # 每个信息是一个元组 (事件相对时间, 事件类型)
    events_in_window = []

    # 遍历每一列（每一种事件类型）
    for event_type in trial_row.index:
        # 我们只关注这些类型的事件
        if event_type in ['LSOUND', 'RSOUND', 'LLICK', 'RLICK', 'LWON_L', 'RWON_L', 'LWOFF_L', 'RWOFF_L', 'NT_END']:
            time_list = trial_row[event_type]
            if time_list: # 如果时间列表非空
                for t_absolute in time_list:
                    # 计算相对时间
                    t_relative = t_absolute - t0
                    # 只保留在时间窗口内的事件
                    if time_window_start <= t_relative <= time_window_end:
                        events_in_window.append((t_relative, event_type))
    
    return events_in_window

# 2. 应用函数到DataFrame的每一行
all_trials_events = []
for idx, row in final_df.iterrows():
    result = align_trial_to_sound(row)
    if result is not None:
        # 将trial编号也存储起来
        for (rel_time, event_type) in result:
            all_trials_events.append({'trial': idx, 'time_rel': rel_time, 'event': event_type})

# 3. 创建新的、对齐后的DataFrame
df_aligned = pd.DataFrame(all_trials_events)
print(df_aligned)
```

```python
from matplotlib.colors import ListedColormap
import numpy as np

# 创建组合事件编码（使用不同的数值代表不同事件类型）
def create_combined_event_data(df_aligned, time_start=0, time_end=3000, bin_size=25):
    """
    创建组合事件的热图数据，将不同事件类型编码为不同的数值
    """
    
    # 1. 【核心修改】先定义时间区间（bins），并据此确定矩阵列数
    time_bins = np.arange(time_start, time_end + bin_size, bin_size)
    n_bins = len(time_bins) - 1  # 时间区间的数量，也是热图矩阵应有的列数
    n_trials = df_aligned['trial'].nunique()  # 试验次数，即热图矩阵的行数

    # 2. 【核心修改】根据计算出的时间区间数量n_bins来初始化矩阵，确保列数匹配
    event_data = np.zeros((n_trials, n_bins))
    
    event_encoding = {
        'LLICK': 1, 'RLICK': 2,
        'LWON_L': 3, 'RWON_L': 3,
        'LWOFF_L': 4, 'RWOFF_L': 4
    }

    for trial_idx in range(n_trials):
        trial_data = df_aligned[df_aligned['trial'] == trial_idx]
        if not trial_data.empty:
            for _, event_row in trial_data.iterrows():
                time_rel = event_row['time_rel']
                event_type = event_row['event']
                
                if event_type in event_encoding:
                    # 使用np.digitize找到时间点对应的bin索引
                    bin_idx = np.digitize(time_rel, time_bins) - 1
                    
                    # 3. 【增强健壮性】严格的边界检查，确保索引在矩阵列数范围内
                    if 0 <= bin_idx < n_bins:  # 现在使用n_bins进行比较
                        event_data[trial_idx, bin_idx] = event_encoding[event_type]
                    # 可选：记录或忽略超出定义时间窗口的事件
                    # else:
                    #    print(f"时间点 {time_rel} 超出窗口，已忽略。")
    
    return event_data, time_bins  # 返回time_bins用于后续坐标轴标注

# 定义参数
time_start = 0
time_end = 3000
bin_size = 25

# 生成数据
combined_data, time_bins_used = create_combined_event_data(df_aligned, time_start, time_end, bin_size)

# 创建自定义颜色映射（组合所有事件）
colors_combined = ['white', 'blue', 'red', 'green', 'purple']
cmap_combined = ListedColormap(colors_combined)

# 绘制图形
plt.figure(figsize=(16, 10))
ax = plt.gca()

# 4. 【重要】设置imshow的extent参数，使其与时间区间对应
# extent的四个参数分别表示X轴起始、X轴结束、Y轴起始、Y轴结束
im = ax.imshow(combined_data, aspect='auto', cmap=cmap_combined,
               extent=[time_start, time_end, combined_data.shape[0], 0])  # 使用time_start和time_end

# ... (其余添加标注线、颜色条、标题的代码保持不变)
sound_events = df_aligned[df_aligned['event'].isin(['LSOUND', 'RSOUND'])]
if not sound_events.empty:
    mean_sound_time = sound_events['time_rel'].mean()
    ax.axvline(x=mean_sound_time, color='black', linestyle='--', 
               linewidth=2, label='Sound Event')

end_events = df_aligned[df_aligned['event'] == 'NT_END']
if not end_events.empty:
    mean_end_time = end_events['time_rel'].mean()
    ax.axvline(x=mean_end_time, color='black', linestyle='-', 
               linewidth=2, label='Trial End')

plt.colorbar(im, label='Event Types\n(1:LLICK, 2:RLICK, 3:Reward, 4:Offset)')
plt.title('Combined Event Heatmap')
plt.ylabel('Trial')
plt.xlabel('Time Relative to Sound Onset (ms)')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from matplotlib.colors import ListedColormap
import matplotlib.patches as mpatches

# 设置图形风格
plt.rcParams['font.sans-serif'] = ['SimHei', 'Arial']  # 支持中文显示
plt.rcParams['axes.unicode_minus'] = False

def create_side_specific_heatmap(df_aligned, side_type, time_start=-1000, time_end=3000, bin_size=10):
    """
    创建侧边特异性热图数据
    side_type: 'left' 或 'right'
    """
    # 定义左右侧事件映射
    event_mapping = {
        'left': {
            'sound': 'LSOUND',
            'lick': 'LLICK', 
            'won': 'LWON_L',
            'off': 'LWOFF_L'
        },
        'right': {
            'sound': 'RSOUND', 
            'lick': 'RLICK',
            'won': 'RWON_L',
            'off': 'RWOFF_L'
        }
    }
    
    events = event_mapping[side_type]
    
    # 创建时间bins
    time_bins = np.arange(time_start, time_end + bin_size, bin_size)
    n_bins = len(time_bins) - 1
    n_trials = df_aligned['trial'].nunique()
    
    # 初始化热图数据矩阵
    heatmap_data = np.zeros((n_trials, n_bins))
    
    # 事件编码
    event_encoding = {
        events['lick']: 1,    # 舔舐事件
        events['won']: 2,     # 奖励事件
        events['off']: 3,     # 偏移事件
        'NT_END': 4          # 试验结束
    }
    
    # 填充热图数据
    for trial_idx in range(n_trials):
        trial_data = df_aligned[df_aligned['trial'] == trial_idx]
        
        for _, event_row in trial_data.iterrows():
            time_rel = event_row['time_rel']
            event_type = event_row['event']
            
            if event_type in event_encoding:
                bin_idx = np.digitize(time_rel, time_bins) - 1
                if 0 <= bin_idx < n_bins:
                    # 只填充当前没有事件的位置
                    if heatmap_data[trial_idx, bin_idx] == 0:
                        heatmap_data[trial_idx, bin_idx] = event_encoding[event_type]
    
    return heatmap_data, time_bins

# 设置参数
time_start = -1000
time_end = 3000
bin_size = 20

# 创建左右侧热图数据
left_heatmap_data, time_bins = create_side_specific_heatmap(
    df_aligned, 'left', time_start, time_end, bin_size
)

right_heatmap_data, _ = create_side_specific_heatmap(
    df_aligned, 'right', time_start, time_end, bin_size
)

# 创建自定义颜色映射
# 0:无事件(白色), 1:舔舐事件, 2:奖励事件, 3:偏移事件, 4:试验结束
colors_left = ['white', 'blue', 'black', 'green', 'black']  # 左侧颜色方案
colors_right = ['white', 'red', 'black', 'green', 'black']  # 右侧颜色方案

cmap_left = ListedColormap(colors_left)
cmap_right = ListedColormap(colors_right)

# 创建图形和子图
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(16, 12))

# 绘制左侧热点图
im1 = ax1.imshow(left_heatmap_data, aspect='auto', cmap=cmap_left,
                vmin=0, vmax=4,  # 固定数值范围
                extent=[time_start, time_end, left_heatmap_data.shape[0], 0],
                interpolation='nearest')  # 禁用插值以获得清晰的色块

ax1.set_title('左侧舔舐行为热点图', fontsize=14, fontweight='bold', pad=20)
ax1.set_xlabel('时间相对于声音开始 (ms)')
ax1.set_ylabel('试验编号')

# 绘制右侧热点图
im2 = ax2.imshow(right_heatmap_data, aspect='auto', cmap=cmap_right,
                vmin=0, vmax=4,
                extent=[time_start, time_end, right_heatmap_data.shape[0], 0],
                interpolation='nearest')

ax2.set_title('右侧舔舐行为热点图', fontsize=14, fontweight='bold', pad=20)
ax2.set_xlabel('时间相对于声音开始 (ms)')
ax2.set_ylabel('试验编号')

# 添加声音起始线（0时刻）
for ax in [ax1, ax2]:
    ax.axvline(x=0, color='gray', linestyle=':', linewidth=1, alpha=0.7)

# 添加事件标记线函数
def add_event_lines(ax, df_aligned, event_types, line_style, color, label_prefix):
    """添加事件标记线"""
    for event_type in event_types:
        event_data = df_aligned[df_aligned['event'] == event_type]
        if not event_data.empty:
            # 按试验分组，计算每个试验中该事件的平均时间
            for trial_idx in df_aligned['trial'].unique():
                trial_events = event_data[event_data['trial'] == trial_idx]
                if not trial_events.empty:
                    mean_time = trial_events['time_rel'].mean()
                    # 在每个试验行上画一条短线
                    ax.plot([mean_time, mean_time], 
                           [trial_idx-0.4, trial_idx+0.4], 
                           color=color, linestyle=line_style, 
                           linewidth=1.5, alpha=0.8)

# 添加事件标记线
# 左侧图：LWON_L（黑色实线），LWOFF_L（绿色实线），NT_END（黑色虚线）
add_event_lines(ax1, df_aligned, ['LWON_L'], '-', 'black', 'WON')
add_event_lines(ax1, df_aligned, ['LWOFF_L'], '-', 'green', 'OFF')
add_event_lines(ax1, df_aligned, ['NT_END'], '--', 'black', 'END')

# 右侧图：RWON_L（黑色实线），RWOFF_L（绿色实线），NT_END（黑色虚线）
add_event_lines(ax2, df_aligned, ['RWON_L'], '-', 'black', 'WON')
add_event_lines(ax2, df_aligned, ['RWOFF_L'], '-', 'green', 'OFF')
add_event_lines(ax2, df_aligned, ['NT_END'], '--', 'black', 'END')

# 创建自定义图例
legend_elements_left = [
    mpatches.Patch(color='white', label='无事件'),
    mpatches.Patch(color='blue', label='左侧舔舐 (LLICK)'),
    plt.Line2D([0], [0], color='black', linestyle='-', linewidth=2, label='奖励事件 (LWON_L)'),
    plt.Line2D([0], [0], color='green', linestyle='-', linewidth=2, label='偏移事件 (LWOFF_L)'),
    plt.Line2D([0], [0], color='black', linestyle='--', linewidth=2, label='试验结束 (NT_END)')
]

legend_elements_right = [
    mpatches.Patch(color='white', label='无事件'),
    mpatches.Patch(color='red', label='右侧舔舐 (RLICK)'),
    plt.Line2D([0], [0], color='black', linestyle='-', linewidth=2, label='奖励事件 (RWON_L)'),
    plt.Line2D([0], [0], color='green', linestyle='-', linewidth=2, label='偏移事件 (RWOFF_L)'),
    plt.Line2D([0], [0], color='black', linestyle='--', linewidth=2, label='试验结束 (NT_END)')
]

ax1.legend(handles=legend_elements_left, loc='upper right', bbox_to_anchor=(1.15, 1))
ax2.legend(handles=legend_elements_right, loc='upper right', bbox_to_anchor=(1.15, 1))

# 调整布局
plt.tight_layout()
plt.show()

# 输出统计信息
print("事件统计信息:")
left_lick_count = len(df_aligned[df_aligned['event'] == 'LLICK'])
right_lick_count = len(df_aligned[df_aligned['event'] == 'RLICK'])
print(f"左侧舔舐次数 (LLICK): {left_lick_count}")
print(f"右侧舔舐次数 (RLICK): {right_lick_count}")
```

```python
# 使用 final_df 的绝对时间绘制：每个事件类型对应一条水平行（x = 绝对时间，y = 事件类别）
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# 事件顺序：将右侧事件放上方，左侧事件放下方
event_order = ['RSOUND','RLICK','RWON_L','RWOFF_L','NT_START','NT_END','LSOUND','LLICK','LWON_L','LWOFF_L']
# 仅保留 final_df 中实际存在的列
event_order = [e for e in event_order if e in final_df.columns]

# 收集每个事件的绝对时间（不考虑 trial 作为 y）
rows = []
for ev in event_order:
    for _, row in final_df.iterrows():
        vals = row.get(ev)
        if isinstance(vals, (list, tuple)) and len(vals) > 0:
            for t in vals:
                rows.append({'event': ev, 'time': t})

pts = pd.DataFrame(rows)

if pts.empty:
    print('没有可绘制的数据（final_df 中未找到事件时间）')
else:
    n = len(event_order)
    # 映射每个事件到一个固定的 y 值：让 event_order 的第一个元素位于最上方
    y_map = {ev: n - 1 - i for i, ev in enumerate(event_order)}

    palette = sns.color_palette('tab10' if n <= 10 else 'tab20', n_colors=n)
    color_map = dict(zip(event_order, palette))

    # 调整图像尺寸、分辨率，横向拉长以显示更多时间跨度；不做垂直抖动，保持每种事件为单列
    plt.figure(figsize=(28, 8), dpi=150)
    ax = plt.gca()

    for ev in event_order:
        sub = pts[pts['event'] == ev]
        if sub.empty:
            continue
        # 保持每个事件的 y 值不变（单列显示），不使用抖动
        y = np.full(len(sub), y_map[ev])
        ax.scatter(sub['time'], y, s=36, color=color_map[ev], label=ev, alpha=0.95,
                   edgecolors='k', linewidths=0.3, zorder=3)

    # 设置 y 轴刻度为事件名称（并按 event_order 从上到下显示）
    ax.set_yticks([y_map[ev] for ev in event_order])
    ax.set_yticklabels(event_order)

    # x轴范围自动设置，并加一点边距（增大到5%以让点分布更宽）
    xmin, xmax = pts['time'].min(), pts['time'].max()
    span = max(1, xmax - xmin)
    x_min_plot, x_max_plot = xmin - 0.05 * span, xmax + 0.05 * span
    ax.set_xlim(x_min_plot, x_max_plot)

    # 在 x 轴上添加适当数量的刻度并格式化显示绝对时间值
    # 使用固定数量的刻度（例如 8 个），并以整数千位分隔符显示，便于观察大数时间戳
    num_xticks = 8
    xticks = np.linspace(x_min_plot, x_max_plot, num=num_xticks)
    ax.set_xticks(xticks)
    ax.ticklabel_format(style='plain', axis='x')  # 关闭科学计数法
    ax.set_xticklabels([f"{int(x):,}" for x in xticks], rotation=45, ha='right', fontsize=10)

    # 保持 y 轴范围（无抖动）
    ax.set_ylim(-0.5, n - 0.5)
    ax.set_xlabel('Absolute time (same units as final_df)')
    ax.set_title('Events as parallel rows — same event types share the same y value')
    ax.grid(axis='x', linestyle=':', alpha=0.5)

    # 当点太多时，图例会很长，可以使用 y 轴标签代替；仍提供图例（可注释掉）
    ax.legend(bbox_to_anchor=(1.02, 1), loc='upper left')

    plt.tight_layout()
    plt.show()
```
